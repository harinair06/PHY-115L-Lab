```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit

# GRAPH 1: Inverse Fit (Length vs. Frequency)


# Data from Table 1
l_data = np.array([65, 60, 55, 50, 45, 40, 35, 30, 25])
f_data1 = np.array(
    [182.617, 192.383, 208.008, 227.539, 250.977, 282.227, 321.289, 377.930, 459.961]
)


def inverse_func(l, a, b):
  return (a / l) + b


# Curve drawing
popt1, pcov1 = curve_fit(inverse_func, l_data, f_data1)
a1, b1 = popt1
perr1 = np.sqrt(np.diag(pcov1))

print("=== Inverse Fit Parameters (Length vs Frequency) ===")
print(f"a = {a1:.4f} ± {perr1[0]:.4f}")
print(f"b = {b1:.4f} ± {perr1[1]:.4f}")
print(f"Equation: f(l) = {a1:.2f} / l + {b1:.2f}\n")

# Plotting
l_fit = np.linspace(min(l_data), max(l_data), 200)

plt.figure(figsize=(7, 4.5))
plt.plot(l_data, f_data1, 'o', color='black', label='Data Points')
plt.plot(
    l_fit,
    inverse_func(l_fit, *popt1),
    color='blue',
    label=f'Fit: $f(l) = \\frac{{{a1:.2f}}}{{l}} + {b1:.2f}$',
)
plt.xlabel('Length, l (cm)', fontsize=12)
plt.ylabel('Frequency, f (Hz)', fontsize=12)
plt.title('Fundamental Frequency vs. Length (Inverse Fit)')
plt.legend()
plt.grid(True)
plt.show()

# GRAPH 2: Square Root Fit (Tension vs. Frequency)

# Data from Table 2
m_data = np.array([1,1.5,2,2.5,3,3.5,4,4.5,5,5.5,6,6.5,7,7.5,8,8.5,9,])
f_data2 = np.array([
    213.867,
    252.930,
    303.711,
    321.289,
    348.633,
    375.977,
    395.508,
    424.805,
    444.336,
    471.680,
    493.164,
    545.898,
    561.523,
    581.055,
    598.633,
    620.117,
    639.648,
])

def sqrt_func(m, a, b):
  return a * np.sqrt(m) + b

# Curve drawing
popt2, pcov2 = curve_fit(sqrt_func, m_data, f_data2)
a2, b2 = popt2
perr2 = np.sqrt(np.diag(pcov2))

print("=== Square Root Fit Parameters (Tension vs Frequency) ===")
print(f"a = {a2:.4f} ± {perr2[0]:.4f}")
print(f"b = {b2:.4f} ± {perr2[1]:.4f}")
print(f"Equation: f(m) = {a2:.2f} * sqrt(m) + {b2:.2f}\n")

# Plotting
m_fit = np.linspace(min(m_data), max(m_data), 200)

plt.figure(figsize=(7, 4.5))
plt.plot(m_data, f_data2, 'o', color='black', label='Data Points')
plt.plot(
    m_fit,
    sqrt_func(m_fit, *popt2),
    color='red',
    label=f'Fit: $f(m) = {a2:.2f}\\sqrt{{m}} + {b2:.2f}$',
)
plt.xlabel('Tension, m (kg)', fontsize=12)
plt.ylabel('Frequency, f (Hz)', fontsize=12)
plt.title('Fundamental Frequency vs. Tension (Square Root Fit)')
plt.legend()
plt.grid(True)
plt.show()



```
